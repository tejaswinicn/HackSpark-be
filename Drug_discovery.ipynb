```python
# Step 1: Load libraries and dataset
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('C:/Users/tejas/Downloads/updated_indian_medicine_data.csv')

# Display shape and column names
print(f"Dataset shape: {df.shape}")
print("Columns:", df.columns.tolist())

# Step 1a: Remove duplicates (if any)
df.drop_duplicates(inplace=True)

# Step 1b: Handle missing values — check nulls
print("\nMissing values:\n", df.isnull().sum())

# Step 1c: Fill or handle missing values (for simplicity — fill NaN texts with empty string)
text_columns = ['short_composition1', 'short_composition2', 'salt_composition',
                'medicine_desc', 'side_effects']
for col in text_columns:
    df[col] = df[col].fillna('')
    
df = df.drop(df.columns[2:7], axis=1)
df = df.drop(df.columns[0], axis=1)
print(df.iloc[0])
```

```python
import pandas as pd
import random
from collections import Counter

# Step 1: Randomly select N drugs
num_drugs_to_mix = 3
selected_rows = df.sample(n=num_drugs_to_mix).reset_index(drop=True)

print("🧪 Selected Drugs for Mixing:")
print(selected_rows[['name', 'short_composition1', 'short_composition2', 'salt_composition']])

# Step 2: Combine components
combined_short_comp1 = ' + '.join(selected_rows['short_composition1'].unique())
combined_short_comp2 = ' + '.join(selected_rows['short_composition2'].unique())
combined_salt_comp = ' + '.join(selected_rows['salt_composition'].unique())

# Synthetic drug name (combine first syllables + random suffix)
name_parts = [name.split()[0][:4] for name in selected_rows['name']]
synthetic_drug_name = ''.join(name_parts) + random.choice(['-X', '-New', '-Plus', '-Max', '-Gen'])

# Step 3: Combine side effects
combined_side_effects = ' | '.join(selected_rows['side_effects'].unique())

# Step 4: Infer potential problem (based on most common words in medicine_desc)
all_desc = ' '.join(selected_rows['medicine_desc'])
words = [word.lower() for word in all_desc.split() if len(word) > 3]
common_words = Counter(words).most_common(5)
predicted_use_case_keywords = [word for word, count in common_words]

# Step 5: Calculate relevance (accuracy proxy)
problem_overlap_count = sum(
    any(kw in desc.lower() for kw in predicted_use_case_keywords) for desc in selected_rows['medicine_desc']
)
relevance_score = round((problem_overlap_count / num_drugs_to_mix) * 100, 2)

# Step 6: Print Final Output
print("\n💊 --- New Synthetic Drug Details ---")
print(f"Name: {synthetic_drug_name}")
print(f"Short Composition 1: {combined_short_comp1}")
print(f"Short Composition 2: {combined_short_comp2}")
print(f"Salt Composition: {combined_salt_comp}")
print(f"\n🩺 Potential Use Case Keywords (inferred): {predicted_use_case_keywords}")
print(f"\n⚠️ Possible Side Effects: {combined_side_effects}")
print(f"\n✅ Relevance Score (accuracy proxy): {relevance_score}%")
```
